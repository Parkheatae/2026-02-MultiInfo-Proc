```python
# 공통부분이기 때문에 실습을 새로운 notebook에서 할때마다 복사해서 실행
import sys
sys.path.append('../multi_info/common') # 모듈이 존재하는 path를 추가
import cvnb
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("../multi_info/images/scaling.jpg", cv2.IMREAD_GRAYSCALE)
if image is None: raise Exception("영상 파일 오류")
print(image.shape)
cv2.imshow("Original", image)
```

```python
def scaling(img, size): # size(w,h)
    dst = np.zeros(size[::-1], img.dtype)
    ratioY, ratioX = np.divide(size[::-1], img.shape[:2]) # size[::-1] = [h, w] 가 출력
    x = np.arange(img.shape[1])
    y = np.arange(img.shape[0])
    y, x = np.meshgrid(y,x)
    j, i = np.int32(y * ratioY), np.int32(x * ratioX)
    dst[j, i] = img[y, x]
    return dst
```

```python
small = scaling(image, (125, 150))
large = scaling(image, (375, 450))
cv2.imshow("Small", small)
cv2.imshow("Large", large)
```

```python
def scaling(img, size): # size(w,h)
    dst = np.zeros(size[::-1], img.dtype)
    ratioY, ratioX = np.divide(size[::-1], img.shape[:2]) # size[::-1] = [h, w] 가 출력
    x = np.arange(img.shape[1])
    y = np.arange(img.shape[0])
    y, x = np.meshgrid(y,x)
    j, i = np.int32(y * ratioY), np.int32(x * ratioX)
    dst[j, i] = img[y, x]
    return dst
```

```python
small = scaling(image, (125, 150))
large = scaling(image, (375, 450))
cv2.imshow("Small", small)
cv2.imshow("Large", large)
```

```python
plt.figure(figsize=(6,3))
plt.subplot(1, 2, 1)
plt.xlim(0, 375)
plt.ylim(450,0)
plt.imshow(small, cmap='gray')
plt.title("Small")
plt.subplot(1,2,2)
plt.imshow(large, cmap='gray')
plt.xlim(0, 375)
plt.ylim(450,0)
plt.title("Large")
plt.show()
```

```python
def scaling_rev(img, size): # size(w,h)
    dst = np.zeros(size[::-1], img.dtype)
    ratioY, ratioX = np.divide(size[::-1], img.shape[:2]) # size[::-1] = [h, w] 가 출력
    i = np.arange(size[0])
    j = np.arange(size[1])
    j, i = np.meshgrid(j,i)
    y, x = np.int32(j / ratioY), np.int32(i / ratioX)
    dst[j, i] = img[y, x]
    return dst
```

```python
small2 = scaling_rev(image, (125, 150))
large2 = scaling_rev(image, (375, 450))
cv2.imshow("Small", small2)
cv2.imshow("Large", large2)
```

```python
plt.figure(figsize=(6,3))
plt.subplot(1, 2, 1)
plt.xlim(0, 375)
plt.ylim(450,0)
plt.imshow(small2, cmap='gray')
plt.title("Small")
plt.subplot(1,2,2)
plt.imshow(large2, cmap='gray')
plt.xlim(0, 375)
plt.ylim(450,0)
plt.title("Large")
plt.show()
```
