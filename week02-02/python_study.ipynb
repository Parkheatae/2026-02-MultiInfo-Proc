## 컬렉션 예제 실행

```python
list1 = [5, 2, 3, 4]
list2 = [1, 1.5, 'a', 'b', '문자열']
tuple1 = (1, 2)
dict1 = {'name': '배종욱', 'email': 'naver.com'}

set1 = {1, '문자열', '2', 'a'}
set2 = {1, '문자열', '2.5', 'b'}

print('list1 =', list1)
print('list2 =', list2)
print('tuple1 =', tuple1)
print('dict1 =', dict1)
print('set1 =', set1)
print('set2 =', set2)

print('insersection =', set1 & set2)  # 교집합
print('union =', set1 | set2)  # 합집합
print('difference =', set1 - set2)  # 차집합
print('symmetric difference =', set1 ^ set2)  # 대칭차집합
print('difference =', set2 - set1)  # 차집합
```
