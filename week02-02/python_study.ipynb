- ### 리스트 자료형과 기본 연산

```python
# 리스트와 튜플의 비교
list1 = [1,2,3,4]
print(list1, type(list1))
list1[0] = 100
print(list1, type(list1))

tuple1 = (1,2,3,4)
print(tuple1, type(tuple1))
tuple1 = list(tuple1)  # 튜플을 리스트로 변환
tuple1[0] = 100  # 튜플은 수정 불가, 에러 발생
print(tuple1, type(tuple1))
```

```python
# 흔히 하는 실수
list = [1,2,3,4]
del list
print(list, type(list))

tuple = (1,2,3,4)
print(tuple, type(tuple))
tuple = list(tuple)  # 튜플을 리스트로 변환
tuple[0] = 100  # 튜플은 수정 불가, 에러 발생
```
