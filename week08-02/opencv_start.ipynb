```python
import numpy as np
import cv2

b, g, r = (255, 0, 0), (0, 255, 0), (0,0,255)
# image = np.zeros((480, 640, 3), np.uint8) # 640x480 color image를 생성하는데, data type unsigned int 8 bits
image = cv2.imread("../multi_info/images/dog.jpg")
print(image.shape)
#image = image.transpose()
#image[:] = (255, 255, 255)
pt1, pt2 = (50, 50), (250, 150)
pt3, pt4 = (400, 150), (500, 50)
roi = (50, 200, 200, 100) # Region Of Interest

cv2.line(image, pt1, pt2, r)
cv2.line(image, pt3, pt4, g, 3, cv2.LINE_AA ) # AA means Anti-Aliasing

cv2.rectangle(image, pt1, pt2, b, 3, cv2.LINE_4)
cv2.rectangle(image, roi, r, 3, cv2.LINE_8)
cv2.rectangle(image, roi, g, -1)

cv2.imshow("blank", image)
```

```python
import numpy as np
import cv2

b, g, r = (255, 0, 0), (0, 255, 0), (0,0,255)
# image = np.zeros((480, 640, 3), np.uint8) # 640x480 color image를 생성하는데, data type unsigned int 8 bits
image = cv2.imread("../multi_info/images/dog.jpg")
print(image.shape)
#image = image.transpose()
#image[:] = (255, 255, 255)
pt1, pt2 = (50, 50), (250, 150)
pt3, pt4 = (400, 150), (500, 50)
roi = (50, 200, 200, 100) # Region Of Interest

cv2.line(image, pt1, pt2, r)
cv2.line(image, pt3, pt4, g, 3, cv2.LINE_AA ) # AA means Anti-Aliasing

cv2.rectangle(image, pt1, pt2, b, 3, cv2.LINE_4)
cv2.rectangle(image, roi, r, 3, cv2.LINE_8)
cv2.rectangle(image, roi, g, -1)

cv2.circle(image, (150, 250), 50, b, -1)

cv2.ellipse(image, (150, 100), (50, 25), 0, 0, 360, r, 4)
cv2.ellipse(image, (150, 100), (50, 25), 45, 0, 360, (255, 255, 0), 4)

cv2.imshow("blank", image)
```
