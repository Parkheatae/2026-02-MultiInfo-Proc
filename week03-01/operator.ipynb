## while 반복문

```python
n = 4  # 4회안에 원하는 값을 입력

while n > 0:
    m = input("Enter a number: ")
    print("You entered: ", m, " ", type(m))
    m = int(m)  # 정수로 변환
    if m == 0: break
    n -= 1  # n-- 는 지원하지 않음. n -= 1 로 대체
else:
    print("You have entered ", 4-n, " numbers without entering 0.")
    print(f"You have entered {4-n} numbers without entering 0.")  # f-string
    print("You have entered {} numbers without entering 0.".format(4-n))  # format string
    print("You have entered %d numbers without entering 0." % (4-n))  # old style string formatting
```

## for 반복문

```python
# range (시작, 종료, 증감) 함수
print(range(0,10))
print(list(range(0,10)))
print(tuple(range(0,10)))

# enumerate() 함수
print(enumerate(range(0,10)))
print(list(enumerate(range(0,20, 2))))
```

```python
kor = ['일', '이', '삼', '사', '오', '육', '칠', '팔', '구', '십']
eng = ['one', 'two', 'three', 'four', 'five', 'six', 'seven', 'eight', 'nine', 'ten']

print(list(zip(kor, eng)))

for (k, e) in zip(kor, eng):
    print(f"{k} = {e}")
```

```python
for i in range(10):
    print(f"{kor[i]} = {eng[i]}")
```

```python
for i in range(0, 10):
    print(f"{kor[i]} = {eng[i]}")
```

```python
for i in range(0, 10, 2):
    print(f"{kor[i]} = {eng[i]}")
```

```python
for i in reversed(range(10)):
    print(f"{kor[i]} = {eng[i]}")
```
