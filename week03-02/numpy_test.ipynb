## numpy 모듈의 학습

```python
import numpy as np
```

```python
list1 = [1,2,3]
list2 = [4.0, 5.0, 6.0]
print(list1, list2)
```

```python
a, b = np.array(list1, dtype=int), np.array(list2, dtype=np.int32)
print(a, b, type(a), type(b))
print('a 원소 데이터 타입:', type(a[0]))
print('b 원소 데이터 타입:', type(b[0]))
```

```python
a-b
```

```python
a*b  # element-wise 연산
```

```python
a/b  # element-wise 나눗셈
```

```python
b//a # 정수 나눗셈
```

```python
print(f"a shape: {np.shape(a)}, b shape: {np.shape(b)}")
```

```python
print(a*3, b*4) #bloadcasting
```

## 무작위 수 (random num) 생성 및 변형

```python
a = np.random.rand(2,3)
print(a, np.shape(a), a.shape)
```

```python
b = np.random.randn(2,3)
print(b)
```

```python
c = np.random.rand(6)
print(c, c.shape)

d = np.random.rand(1,6)
print(d, d.shape)
```

```python
e = np.random.randint(0, 101, 10)
print(e, e.shape)
```

## 형태(모양) 변환

```python
print(a, a.shape)
c3 = a.reshape(-1)  # view 형식만 변경
print(c3, c3.shape)
c3[0] = 1
print(a)
```

```python
print(a, a.shape)
c4 = a.reshape((3,2))  # 원래 a값을 보존하고, a의 형태 (view 형식)만 c2가 가져감
print(c4, c4.shape)
c4[0][0] = 0.5
print(a)
```
