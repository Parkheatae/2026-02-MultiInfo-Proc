### numpy의 배열 생성 (ones, zeros) : ones, 모든 원소가 1임, zeros, 모든 원소가 0임


```python
a = np.ones((2,3), np.float32)
b = np.ones(10, np.uint8)
print(a, b)
```

```python
print(a.shape, b.shape)
```

```python
b = b.reshape((10,1)) # vector (10,) -> matrix (10, 1) 로 변환
print(a.shape, b.shape)
```

```python
c = np.zeros((1,2,3), np.float32)
print(c.shape, c)
```

```python
a_shape = a.shape
d = np.zeros(a_shape, a.dtype)
print(a.shape, d.shape, a.dtype, d.dtype)
```

```python
e = np.zeros_like(a)
print(e.shape, e.dtype)
```

```python
f = np.arange(2, 5, dtype=np.float32)
print(f)
```

```python
# arange from 1 to 3*3*4, reshape, (3,3,4) 배열을 생성
g = np.arange(1, 3*3*4+1, dtype=np.float32).reshape(3,3,4)
print(g, g.shape, g.ndim)
```

#### 배열의 병합
- 세로 병합 A : (3,5), B: (2, 5) ==> (5,5)
- 가로 병합 A : (5,3), B: (5, 2) ==> (5,5)

```python
a = np.ones((3,5), dtype=np.float32)
b = np.zeros((2,5), dtype=np.float32)
print(a, b)
```

```python
c = np.concatenate([a, b])
print(c, c.shape)
```

```python
d = np.concatenate([a.T,b.T], axis=1)
print(d.shape, d)
```

### view vs copy

### 행렬 연산

```python
a = np.array([[2,3,4],
              [5,6,7]], dtype=np.float32)

b = np.array([[20,21,22],
              [23,24,25]], dtype=np.float32)
```

```python
print(a+b, a*b)
```

```python
c = np.dot(a, b.T)
print(c, c.shape)

d = np.matmul(a, b.T)
print(d, d.shape)
```

```python
e = d>300
print(e)
d*e
```

```python
c = np.dot(a, b.T)
print(c, c.shape)

d = np.matmul(a, b.T)
print(d, d.shape)
```

```python
e = d>300 # 조건 검사
print(e) # boolean 행렬 출력
print(d*e) # index가 True인 원소 값만 보존, 나머지는 0으로 치환
print(d[e]) # index가 True인 원소만 출력
print(d[e].astype(np.int32)) # data type을 변환
```
